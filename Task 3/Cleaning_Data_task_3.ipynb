**1. Data Inspection**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("retail_store_sales task 3.csv")

# Display first 5 rows
df.head()
```

```python
# Check dataset size
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

**2. Handling Missing Values**

```python
# Calculate missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
# Handle missing values

# Categorical column
df['Item'] = df['Item'].fillna(df['Item'].mode()[0])

# Numerical columns
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Price Per Unit'].median())
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['Total Spent'] = df['Total Spent'].fillna(df['Total Spent'].median())

# Discount Applied
df['Discount Applied'] = df['Discount Applied'].fillna(
    df['Discount Applied'].mode()[0]
)
```

```python
# Check missing values after handling
df.isnull().sum()
```

**3. Duplicate Records**

```python
# Check for duplicate rows
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
# Display duplicate rows
df[df.duplicated()]
```

**4. Handling Inconsistent Data**

```python
# Check unique values in categorical columns

print("Category:")
print(df['Category'].unique())

print("\nPayment Method:")
print(df['Payment Method'].unique())

print("\nLocation:")
print(df['Location'].unique())

print("\nDiscount Applied:")
print(df['Discount Applied'].unique())
```

```python
print("Number of unique items:", df['Item'].nunique())
print(df['Item'].unique())
```

```python
# Check unique values of categorical columns

print("Category:")
print(df['Category'].unique())

print("\nPayment Method:")
print(df['Payment Method'].unique())

print("\nLocation:")
print(df['Location'].unique())

print("\nDiscount Applied:")
print(df['Discount Applied'].unique())
```

```python
# Remove leading and trailing spaces from text columns

text_columns = ['Category', 'Item', 'Payment Method', 'Location']

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
# Check categorical values after standardization

for col in text_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

**5. Correcting Data Types**

```python
# Check current data types
df.dtypes
```

```python
# Convert Transaction Date to datetime format
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])

# Check data types again
df.dtypes
```

**6. Outlier Detection**

```python
# Check outliers using IQR method

numerical_columns = ['Price Per Unit', 'Quantity', 'Total Spent']

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"{col}:")
    print(f"Number of outliers: {len(outliers)}")
    print(f"Lower Bound: {lower_bound}")
    print(f"Upper Bound: {upper_bound}")
    print()
```

```python
# View potential outliers in Total Spent

Q1 = df['Total Spent'].quantile(0.25)
Q3 = df['Total Spent'].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR

total_spent_outliers = df[df['Total Spent'] > upper_bound]

total_spent_outliers[['Transaction ID', 'Category', 'Item',
                      'Price Per Unit', 'Quantity', 'Total Spent']].head(20)
```

```python
print("Maximum Total Spent:", df['Total Spent'].max())
```

**7. Normalization / Standardization**

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Price_Normalized', 'Quantity_Normalized', 'Total_Spent_Normalized']] = scaler.fit_transform(
    df[['Price Per Unit', 'Quantity', 'Total Spent']]
)

df[['Price Per Unit', 'Price_Normalized',
    'Quantity', 'Quantity_Normalized',
    'Total Spent', 'Total_Spent_Normalized']].head()
```

**8. Feature Engineering**

```python
df['Year'] = df['Transaction Date'].dt.year
df['Month'] = df['Transaction Date'].dt.month
df['Day'] = df['Transaction Date'].dt.day

df[['Transaction Date', 'Year', 'Month', 'Day']].head()
```

```python
df['Amount_Per_Item'] = df['Total Spent'] / df['Quantity']

df[['Total Spent', 'Quantity', 'Amount_Per_Item']].head()
```

**9. Before vs After Data Comparison**

```python
print("Dataset shape after cleaning:", df.shape)
print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())

print("\nData types after cleaning:")
print(df.dtypes)
```

 **10. Save Cleaned Dataset**
